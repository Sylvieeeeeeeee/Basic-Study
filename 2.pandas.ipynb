# 一、Series
## 1.Series的创建

```python
import pandas as pd
s = pd.Series([10,2,3,4,5])
print(s) #左边索引，右边数据
#自定义索引
s = pd.Series([10,2,3,4,5],index=['a','b','c','d','e'])
#定义name
s = pd.Series([10,2,3,4,5],index=['a','b','c','d','e'],name  = '月份')
print(s)
```

```python
# 通过字典来创建
s = pd.Series({'a':10,'b':2,'c':3,'d':4})
print(s)
s1 = pd.Series(s,index = ['a','d']) # 可以只取出一部分，“s“必须要是Series
print(s1)
```

## 2.Series的属性

```python
print(s)
print(s.index)
print(s.values)
print(s.shape,s.ndim,s.size)
print(s.dtype)
s.name = "test"
print(s.name)
print(s.loc['a']) #显式索引，用自定义的index获取
print(s.loc['a':'c']) #支持切片
print(s.iloc[0:2]) #隐式索引，用默认的index获取
print(s.at['a'])
print(s.iat[0]) # at 和 iat 都不支持切片
```

## 3.访问数据

```python
import pandas as pd
s = pd.Series([4,2,3,1,5])
print(s[0])
s = pd.Series([4,2,3,1,5],index=['a','b','c','d','e'])
print(s['a'])
#print(s[0]) # 如果有用自定义标签的话，不能通过默认标签的索引访问数据
print(s[s<3]) #布尔索引，中括号里写对value的布尔表达式
s['f'] = 6 #可以随时加一个，跟字典一样
print(s.head()) # 默认打印前五行
print(s.head(2)) #前两行
print(s.tail()) # 默认打印后五行
print(s.tail(3)) #后三行
```

# 二、DataFrame

## 1.dataframe 的创建方式

```python
import pandas as pd
import numpy as np

# 通过Series来创建
s1 =pd.Series([1,2,3,4,5],index=['a','b','c','d','e'])
s2 =pd.Series([6,7,8,9,10]) # 因为s2中没有index为abcde的值，所以dataframe里面s2的值全部都是缺失值
df = pd.DataFrame({"第一列":s1,"第二列":s2})
print(df)

# 通过字典来创建
df = pd.DataFrame({
    "id":[1,2,3,4,5],
    "name":["a","b","c","d","e"],
    "age":[20,20,30,40,50]
},index = ['a','b','c','d','e'],columns = ['name','age']) # 可以调整参数的顺序，也可以只获取部分参数
print(df)
```

## 2.dataframe的属性

```python
print('行索引：')
print(df.index)
print('列标签：')
print(df.columns)
```

```python
print('值：')
print(df.values) # 返回一个矩阵
```

```python
print('维度：',df.ndim)
print('形状',df.shape)
print('元素个数',df.size)
print('数据类型：',df.dtypes)
```

```python
print('行列转置：')
#print(df.T)
print(df.T.columns) # 行索引和列标签也进行了变化
```

```python
# 获取元素 loc iloc at iat
#获取某行数据
print(df.loc['b'])
print(df.iloc[2])
```

```python
# 获取某列数据
print(df.loc[:,'name'])
print(df.iloc[:,0]) #切片方法
```

```python
# 单个数据
print(df.at['b','name'])
print(df.iat[1,0])
print(df.iloc[1,0])
```

## 3.获取数据

```python
print(df['name'])
print(df.name)
print(df.loc[:,'name'])
```

```python
print(df[['name','age']]) #返回值是dataframe，可以获取多列值
```

```python
# 查看部分数据
print(df.head(2))
print(df.tail(2))
```

```python
# 布尔索引
df.age>10 # 只能判断
```

```python
df[df.age>10] # 获取全部信息
```

```python
df[(df.age>10) & (df.name>'b')] # 两个条件需要用（）括起来
```

```python
# 随机抽样
df.sample(3) # 随机取多条数据
```

## 4.常用方法

```python
df['score'] = [66,77,68,99,87] # loc 赋值
#df.drop(columns = [2]) # 会保留原来列，只删除数据
#del df[2] # 直接删除该列，完全删除
del df['socre']
df
```

```python
print(df.head(3))
print(df.tail(3))
```

```python
print(df.isin([66,87,'a','c']))
```

```python
print(df.isna())
```

```python
print(df.sum()) # 每一column：字符串拼接，数字相加
print(df['score'].sum())
print(df.score.max())
# .mean()
# .median() 中位数
# .mode()
# .quantile(q) 分位数
# .std() 标准差
# .var() 方差
```

```python
print(df.describe())
```

```python
df.count()
```

```python
df.value_counts() # 只有完全一样的一行才可以算作两次
```

```python
# 去重
# df.drop_duplicates()
# 检查是否是重复的
# df.duplicated()
# 可以指定列查看是否是重复的
df.duplicated(subset='age')
```

```python
df.sample()
```

```python
df.replace(50,30) # 把所有的50换成30,也可以指定列操作
```

```python
# 累计和累计积
df.cumsum() #可以用参数axis改变按行按列：0-->按列 ； 1-->按行
df.cummin() #累计出现的最小的，和df.max()效果一样
```

```python
# 排序
# 按索引排序
print(df.sort_index()) #默认从低到高排序
print(df.sort_index(ascending=False)) #可改变参数为降序排列
```

```python
# 按值排序
print(df.sort_values(by='age')) # 参数by= ， 按照什么排序
print(df.sort_values(by=['age','name'], ascending=[True,False])) # 可以设置多个参照，并且每个都可以独立设置升序和降序
```

```python
# 取最大的两个
df.nlargest(2,columns = 'age') # 同样可以写多列 ： columns = ['age','name','score'],也可以改升序降序
```

```python
# 取最小的两个
df.nsmallest(2,columns = 'age')
```
