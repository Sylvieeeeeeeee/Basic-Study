## 2.数据处理

```python
# 缺失值的处理
# np.nan：not a number
# None,pd.NA:don't exist 不存在
import pandas as pd
import numpy as np
s = pd.Series([1,2,np.nan,None,pd.NA])
print(s)
df = pd.DataFrame([[1,pd.NA,2],[2,3,4],[np.nan,None,3]])
df
# 查看是否是缺失值
print(s.isna())
print(s.isnull())
print(s.isnull().sum())
print(df.isna())
print(df.isnull())
print(df.isnull().sum()) #.sum 查看每个column的缺失值个数
```

```python
# 剔除缺失值
print(s.dropna())
print(df)
print(df.dropna()) # 会删除有缺失值的那一整行
print(df.dropna(how='any'))
print(df.dropna(how='all')) # 这一整行全是na才删除
print(df.dropna(thresh=2)) # 如果至少有n个值不是na，就保留
print(df.dropna(axis=1)) #剔除有na的那一整列
print(df.dropna(subset=[1])) #column=1的地方有缺失值就删掉，可以用来指定筛查某column
```

```python
# 填充缺失值
df = pd.read_csv('data/student_scores .csv')
df
print(df.isna().sum())
print(df.fillna({'语文':40})) # 使用字典来填充，固定填充缺失值：df.fillna({'column':固定值})
print(df.fillna(df[['数学']].mean())) # 数学行用平均值来填充：df.fillna(df[['column']].mean()
# 关于[[]] 和 [] ,前者返回一个只有一列的dataframe，后者返回一个数，建议用[],同时选两个columns的时候再用[[]]
print(df.ffill()) # 用前面的值填充
print(df.bfill()) # 用后面的值填充
```

```python
# 处理重复数据
data = {
    "name":['alice','alice','bob','alice','jack','bob'],
    "age":[26,25,30,25,35,30],
    "city":['NY','NY','LA','NY','SF','LA']
}
df = pd.DataFrame(data)
df
```

```python
df.duplicated() # 一整条记录都相同，标记为重复
df.drop_duplicates() # 直接把重复的那一整条删掉
print(df.drop_duplicates(subset=['name']) )# 可以指定删除哪特定column有重复的那一条数据
df.drop_duplicates(subset=['name'], keep='last') # keep='last':保留最新的那一条数据
```

```python
df['月薪'] = df['月薪'].astype('int16') #优化数据：int64对于某些数据太大了，永远都用不完。可以换成范围小一点的
df.dtypes
```

```python
# 数据变形
data = {
    'ID':[1,2],
    'name':['alice','bob'],
    'math':[90,85],
    'English':[88,92],
    'Science':[95,89]
}
df = pd.DataFrame(data)
df
df.T # 最简单的变形：转置
```

```python
'''
ID name 科目
1 alice math 90
1 alice English 88
1 alice Science 90
'''
# 宽表转换成长表
df2 = pd.melt(df,id_vars=['ID','name'],var_name='科目',value_name='分数') # 新建了一个表，id_vars=['不变的列'],var_name='新建的列’，value_name='值的名字'
df2.sort_values(['name'])
```

```python
# 长表转宽表
pd.pivot(df2,index=['ID','name'],columns='科目',values='分数') # index=['不变的列'],columns='用哪一列当列名',value：列里面的数据值
```

```python
# 数据分箱 pd.cut(x,bins,labels): x-->df,series;bins-->区间;labels-->每个区间叫什么，数据类型为list
df = pd.read_csv('data/employees .csv',encoding='gbk')
df.head()
df1 = df.head(10)[['工号','月薪']]
df1
```

```python
pd.cut(df1['月薪'],bins=3) # bins=n,分成n段区间
pd.cut(df1['月薪'],bins=3).value_counts()
pd.cut(df1['月薪'],bins=[0,10000,20000,30000,40000]).value_counts() # bins也可以传入list作为参数
df1['经济水平'] = pd.cut(df1['月薪'],bins=[0,10000,20000,30000,40000],labels=['poor','low','medium','high'])
df1
```

```python
# 等人数分箱
pd.qcut(df1['月薪'],3)
pd.qcut(df1['月薪'],3).value_counts()
```
